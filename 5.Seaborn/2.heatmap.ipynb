```python
import seaborn as sns
from matplotlib import pyplot as plt
```

## Heatmap Basic

```python
sns.get_dataset_names()
```

```python
data = sns.load_dataset('tips')
data
```

```python
data.info()
```

```python
# dataset er all numerical column gular correlation ber korar jonne data.corr() use kora hoy

corr = data.corr(numeric_only = True)
corr
```

```python
# color joto dark correlation toto kom
sns.heatmap(data = corr)
```

```python
# graph er sathe value show korar jonne annot = True use kora hoy
sns.heatmap(data = corr, annot = True)
```

## Heatmap with a large dataset

```python
sns.get_dataset_names()
```

```python
data = sns.load_dataset('flights')

data.head()
```

```python
# ekhane 144 ta row ase and 3 ta column ase.
data.info()
```

```python
# ekhon ami ei full data ta ke pivot table e convert korbo.
# pivot table holo emon ekta table jeta 2 ta cateogorical data ke row and column e convert kore and tar sapekkhe numerical column er value ke represent kore.


pivot = data.pivot_table(index = 'month', columns = 'year', values = 'passengers')
pivot
```

```python
# ekhon ei pivot table ke heatmap e convert korbo.

sns.heatmap(data = pivot)
```

```python
sns.heatmap(data = pivot, cmap = 'coolwarm')
```

```python
sns.heatmap(data = pivot, cmap = 'coolwarm', linecolor = 'white', linewidth = 1)
```
