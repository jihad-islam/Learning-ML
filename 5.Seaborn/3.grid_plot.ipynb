```python
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
data = sns.load_dataset("tips")
data.head()
```

## PairGrid Plot

```python
# amra aage pair plot dekhesi. grid plot similar to pair plot. but grid plot is more customizable.
# graph er output er text e dekho pair plot ta ekta PairGrid.

sns.pairplot(data = data)
```

```python
grid = sns.PairGrid(data = data)
```

```python
# grid class er moddhe by default ekta map method pawa jay. map method er moddhe amra kono ekta plot type dite pari. jemon scatter, hist, kde etc.

grid = sns.PairGrid(data = data)
grid.map(sns.scatterplot)
```

```python
# amra grid plot e etaw define kore dite pari ami just diagonaly plot dekhbo naki upper triangle e plot dekhbo naki lower triangle e plot dekhbo. amra chaile upper triangle e scatterplot r lower triangle e kde plot dite pari. mane onek customization kora jay.

# fill diagonally
grid = sns.PairGrid(data = data)
grid.map_diag(sns.histplot)
```

```python
# fill upper triangle
grid = sns.PairGrid(data = data)
grid.map_upper(sns.scatterplot)
```

```python
# ekhn diagonally ek type er plot, and upper triangle e ek type er plot, and lower triangle e ek type er plot dite pari.

grid = sns.PairGrid(data = data)
grid.map_diag(sns.histplot)
grid.map_upper(sns.scatterplot)
grid.map_lower(sns.kdeplot)
```

```python
grid = sns.PairGrid(data = data)
grid.map_diag(sns.histplot)
grid.map_upper(sns.scatterplot)
grid.map_lower(sns.kdeplot, fill = True)
```

```python
grid = sns.PairGrid(data = data, hue = "sex", palette = "coolwarm")
grid.map_diag(sns.histplot)
grid.map_upper(sns.scatterplot)
grid.map_lower(sns.kdeplot, fill = True)
```

## FacetGrid()

```python
# FacetGrid is a general way to create a grid of plots based on the values of one or more categorical variables. It allows you to create multiple subplots, each representing a subset of the data based on the levels of the categorical variable(s). This is useful for visualizing how different categories relate to each other and to the data.

sns.FacetGrid(
    data = data,
    col = "sex",
    row = 'smoker'
)
```

```python
grid = sns.FacetGrid(
    data = data,
    col = "sex",
    row = 'smoker'
)

grid.map(sns.scatterplot, "total_bill", "tip")
```

```python
sns.set_style("darkgrid")

grid = sns.FacetGrid(
    data = data,
    col = "sex",
    row = 'smoker'
)

grid.map(sns.scatterplot, "total_bill", "tip")
```
