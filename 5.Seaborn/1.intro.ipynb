# Distribution Plot

```python
'''
    Seaborn: Distribution Plot ও Categorical Plot

    ১. Distribution Plot (ডেটার বিস্তৃতি দেখা)
    ------------------------------------------
    কাজ:
    - একটি Numerical Variable-এর ডেটা কীভাবে ছড়ানো আছে (Spread), তা বুঝতে।
    - ডেটার Range এবং কোথায় ডেটা বেশি ঘনীভূত (Density), তা দেখতে।

    প্রধান Plot-গুলো:
    - sns.histplot() -> ডেটাকে ছোট ছোট ভাগে (bins) ভাগ করে count বা সংখ্যা দেখায়।
    - sns.kdeplot()  -> Histogram-এর মতোই, কিন্তু smooth curve দিয়ে density বোঝায়।

    ML Learner-এর Takeaways:
    - ডেটা Normal Distribution-এ আছে নাকি Skewed (ডানে বা বামে হেলে আছে) তা চেক করা।
    - ডেটাতে Outlier বা অস্বাভাবিক ভ্যালু আছে কিনা তা দ্রুত বোঝা।
    - Decision: ডেটা Skewed থাকলে কিছু মডেলের (যেমন Linear Regression) জন্য Transformation (যেমন: Log Transform) দরকার হতে পারে।


    ২. Categorical Plot (ক্যাটাগরি ভিত্তিক বিশ্লেষণ)
    ------------------------------------------------
    কাজ:
    - একটি Categorical Variable-এর (যেমন: day, sex) সাথে Numerical Variable-এর (যেমন: total_bill) সম্পর্ক দেখা।
    - বিভিন্ন গ্রুপের বা ক্যাটাগরির মধ্যে তুলনা করা।

    প্রধান Plot-গুলো:
    - sns.countplot() -> শুধু ক্যাটাগরিগুলোর সংখ্যা গোনার জন্য (যেমন: কতজন Smoker)।
    - sns.barplot()   -> ক্যাটাগরি অনুযায়ী Numerical ভ্যালুর গড় (Mean) দেখায়।
    - sns.boxplot()   -> (খুবই গুরুত্বপূর্ণ!) এটি ডেটার Quartiles, Median এবং Outliers বাক্সের মাধ্যমে দেখায়। Outlier ধরার জন্য বেস্ট।
    - sns.violinplot()-> Boxplot এবং KDE Plot-এর মিশ্রণ, যা বক্সের সাথে ডেটার ঘনত্বও দেখায়।

    ML Learner-এর Takeaways:
    - Target Variable-কে কোন Categorical Feature সবচেয়ে বেশি প্রভাবিত করে, তা বুঝতে পারা।
    - Boxplot ব্যবহার করে প্রতিটি ক্যাটাগরির জন্য আলাদাভাবে Outlier সনাক্ত করা।
    - Decision: Model-এর জন্য কোন Categorical Feature-টি রাখা উচিত (Feature Selection), তার সিগন্যাল পাওয়া।

    -----------------------------------------------------------
    মনে রাখার সহজ উপায়:
    Distribution Plot -> "আমার Numerical ডেটার সামগ্রিক চেহারা কেমন?"
    Categorical Plot  -> "বিভিন্ন ক্যাটাগরি বা গ্রুপের মধ্যে ডেটার পার্থক্য কী?"
'''
```

```python
'''
    Graph theke findings:

    1. total_bill ও tip-এর মধ্যে Positive Relationship আছে
    --------------------------------------------------------
    Observation:
    - total_bill বাড়লে সাধারণত tip-ও বাড়ছে।
    - অর্থাৎ X ↑ হলে Y সাধারণত ↑।

    ML Implication:
    - total_bill একটি potentially useful feature হতে পারে যদি tip predict করতে চাই।
    - অর্থাৎ এটি target-এর জন্য predictive signal দিতে পারে।

    Decision:
    - total_bill-কে candidate feature হিসেবে রাখা যেতে পারে।
    - পরে correlation/model performance দিয়ে verify করতে হবে।

    Important:
    - Relationship থাকলেই feature অবশ্যই useful হবে—এটা ধরে নেওয়া যাবে না।
    - Relationship ≠ Causation।


    2. Relationship আছে, কিন্তু Perfect না
    --------------------------------------
    Observation:
    - Scatter plot-এর points একটি straight line-এর উপর নেই।
    - একই total_bill-এর জন্য বিভিন্ন tip value দেখা যাচ্ছে।

    ML Implication:
    - total_bill একা tip-এর সম্পূর্ণ variation explain করতে পারছে না।
    - অন্য feature-ও tip prediction-এ গুরুত্বপূর্ণ হতে পারে।

    Possible additional features:
    - size
    - sex
    - smoker
    - day
    - time

    Decision:
    - শুধুমাত্র একটি feature-এর উপর নির্ভর না করে অন্য features explore করতে হবে।
    - Feature selection/feature engineering-এর জন্য এই observation useful।


    3. Outlier / Extreme Observation আছে
    -------------------------------------
    Observation:
    - কিছু points main data cluster থেকে অনেক দূরে।
    - যেমন total_bill এবং tip-এর কিছু unusually high value।

    ML Implication:
    - Extreme value কিছু model-এর learning-এ disproportionate influence ফেলতে পারে।
    - বিশেষ করে mean/std-based preprocessing বা কিছু regression approach-এ impact হতে পারে।

    Decision:
    Outlier দেখলেই remove করা যাবে না।

    Process:
        Outlier detected
            ↓
        Validate the observation
            ↓
        Data error নাকি real observation?
            ↓
        Error  -> Correct / Remove
        Real   -> Keep বা suitable treatment

    Possible treatment:
    - Keep
    - Remove
    - Transformation
    - Capping/Clipping
    - Robust method

    Important:
    - "Outlier আছে → remove করতে হবে" — এই rule ভুল।


    4. total_bill-এর Distribution Right-Skewed
    ------------------------------------------
    Observation:
    - Top histogram-এ বেশিরভাগ total_bill lower/middle range-এ।
    - কিছু বড় value ডানদিকে long tail তৈরি করেছে।
    - তাই distribution right-skewed / positively skewed।

    ML Implication:
    - Highly skewed feature কিছু model ও preprocessing method-এ সমস্যা করতে পারে।
    - বিশেষ করে extreme values mean/std-কে influence করতে পারে।

    Decision:
    - Feature transformation দরকার হতে পারে।
    - যেমন log transformation test করা যেতে পারে।

    Important:
    - Skewed হলেই automatically transformation করতে হবে—এটাও ঠিক নয়।
    - Model ও data-এর behavior দেখে decision নিতে হবে।


    5. tip-এর Distribution-ও Right-Skewed
    -------------------------------------
    Observation:
    - বেশিরভাগ tip relatively small value range-এ।
    - কিছু unusually large tip আছে।
    - Distribution-এ right tail আছে।

    ML Implication:
    - Target (`tip`) evenly distributed নয়।
    - Regression model train করার সময় large target values-এর behavior আলাদা হতে পারে।

    Decision:
    - Target distribution বুঝে model-এর prediction ও error behavior evaluate করতে হবে।
    - পরে residual/error analysis useful হবে।


    6. total_bill বাড়ার সাথে tip-এর Spread কিছুটা বাড়ছে
    ------------------------------------------------------
    Observation:
    - ছোট total_bill-এর ক্ষেত্রে tip-এর variation তুলনামূলক কম।
    - বড় total_bill-এর ক্ষেত্রে tip-এর possible values বেশি spread হচ্ছে।

    ML Implication:
    - এটি heteroscedasticity-এর সম্ভাব্য indication হতে পারে।
    - অর্থাৎ X-এর level বদলানোর সাথে Y-এর variability একই রকম থাকছে না।

    Decision:
    - Regression model-এর পরে residual plot/diagnostic check করা useful হতে পারে।
    - শুধু model score দেখে সিদ্ধান্ত না নিয়ে error pattern-ও দেখতে হবে।

    Important:
    - এই graph alone থেকে নিশ্চিত heteroscedasticity বলা যাবে না।
    - এটি শুধু একটি signal।


    7. Relationship চোখে দেখা যাচ্ছে → পরে Measure করতে হবে
    ---------------------------------------------------------
    Observation:
    - Graph দেখে positive relationship দেখা যাচ্ছে।

    ML/Statistics Implication:
    - Visual observation useful, কিন্তু quantitative evidence আরও ভালো।

    Decision:
    - Correlation calculate করা যেতে পারে:

        df["total_bill"].corr(df["tip"])

    Concept:
        Graph     → Pattern detect
        Correlation → Pattern quantify

    অর্থাৎ EDA graph hypothesis তৈরি করে,
    তারপর statistical measure দিয়ে সেটা যাচাই করা হয়।


    8. Feature useful মনে হচ্ছে → Model-এ impact verify করতে হবে
    --------------------------------------------------------------
    Observation:
    - total_bill ও tip-এর মধ্যে relationship আছে।

    Possible ML Decision:
    - total_bill candidate feature হিসেবে রাখা যেতে পারে।

    কিন্তু final decision:
    - Model training-এর পরে evaluate করতে হবে।
    - Feature যোগ করলে prediction performance improve করে কিনা দেখতে হবে।

    Important:
        "Relationship আছে"
            ≠
        "Feature অবশ্যই useful"

    Feature usefulness শেষ পর্যন্ত model/validation-এর মাধ্যমে verify করতে হবে।


    9. এই graph থেকে ML Engineer-এর মূল mindset
    --------------------------------------------
    EDA করার সময় শুধু "কী দেখছি?" জিজ্ঞেস করব না।

    বরং:

        Graph
        ↓
        Observation
        ↓
        কেন এটা গুরুত্বপূর্ণ?
        ↓
        ML-এ কী impact হতে পারে?
        ↓
        কী test/measure করা দরকার?
        ↓
        তারপর Decision


    Example 1:
        Outlier detected
            ↓
        Model-এ influence করতে পারে
            ↓
        Validate
            ↓
        Error নাকি real?
            ↓
        তারপর treatment


    Example 2:
        Positive relationship
            ↓
        Feature predictive হতে পারে
            ↓
        Correlation / model performance check
            ↓
        Useful হলে feature রাখা


    Example 3:
        Skewed distribution
            ↓
        Preprocessing impact হতে পারে
            ↓
        Transformation দরকার কিনা evaluate
            ↓
        Model অনুযায়ী decision


    Final ML takeaway:
    ------------------
    EDA-এর উদ্দেশ্য শুধু graph পড়া নয়।

    EDA-এর উদ্দেশ্য হলো:

        Data-এর pattern খুঁজে বের করা
                ↓
        Problem/Opportunity identify করা
                ↓
        ML-এর উপর সম্ভাব্য impact বোঝা
                ↓
        Test/Validate করা
                ↓
        Evidence-based ML decision নেওয়া


    সবচেয়ে গুরুত্বপূর্ণ:
    - Observation ≠ Final Decision
    - Graph → Hypothesis/Signal দেয়
    - Statistics/Testing/Model Evaluation → Decision validate করতে সাহায্য করে

'''

pass
```

## pairplot

```python
'''
    PairPlot (sns.pairplot)

    কাজ:
    - Dataset-এর সবগুলো Numerical Variable(only numeric columns)-এর মধ্যকার Relationship একসাথে দেখার জন্য ব্যবহার করা হয়।
    - এটি একটি Grid বা Matrix তৈরি করে, যেখানে প্রতিটি Numerical Column একে অপরের সাথে Plot হয়।

    কী কী দেখায়?
    1. Diagonal Plots (কোণাকুণি) -> Single Variable-এর Distribution দেখায় (সাধারণত Histogram বা KDE)।
    2. Off-Diagonal Plots (বাকিগুলো) -> দুটি ভিন্ন Variable-এর Relationship দেখায় (সাধারণত Scatter Plot)।

    কেন ব্যবহার করা হয়?
    - পুরো Dataset-এর Big Picture বা সামগ্রিক চিত্র এক নজরে বুঝতে।
    - কোন কোন Feature-গুলোর মধ্যে Strong Correlation আছে, তা দ্রুত খুঁজে বের করতে।
    - Classification Task-এ Categorical Variable অনুযায়ী Data কতটা আলাদা করা যায় (Separable), তা চেক করতে।

    Common Arguments:
    - hue="categorical_column" -> Categorical Data অনুযায়ী আলাদা রঙ করে (Data Separation দেখার জন্য সবচেয়ে Useful)।
    - kind="scatter" (Default) -> Off-diagonal plot-এর ধরন নির্ধারণ করে।
    - diag_kind="auto" / "hist" / "kde" -> Diagonal plot (Distribution)-এর ধরন নির্ধারণ করে।

    মনে রাখার Shortcut:
    PairPlot = Dataset-এর সব Numerical Column-এর JointPlot-এর বিশাল Grid।

    --------------------------------------------------------
    ML Engineer হিসেবে PairPlot থেকে কী খুঁজব?
    --------------------------------------------------------

    1. Feature বনাম Feature (Multicollinearity):
    - Observation: Off-diagonal plot-এ দুটি Independent Feature-এর মধ্যে Strong Linear Relationship দেখা যাচ্ছে।
    - ML Implication: Linear Model (যেমন Linear Regression)-এ দুটি Highly Correlated Feature রাখলে Model Unstable হতে পারে।
    - Decision: Correlation চেক করে যেকোনো একটি Feature-কে Drop করার কথা ভাবা যেতে পারে (Feature Selection)।

    2. Feature বনাম Target (Predictive Power):
    - Observation: কোনো Feature এবং Target Variable-এর Scatter plot-এ স্পষ্ট Pattern দেখা যাচ্ছে।
    - ML Implication: ওই Feature-টি Target-কে Predict করার জন্য ভালো Signal দিচ্ছে।
    - Decision: এটিকে Model-এর জন্য Candidate Feature হিসেবে রাখা।

    3. Class Separation (Classification Task-এর ক্ষেত্রে):
    - Observation: `hue` ব্যবহার করার পর দেখা গেল বিভিন্ন রঙের Points খুব সুন্দরভাবে আলাদা হয়ে আছে (Linearly Separable)।
    - ML Implication: Classification Model (যেমন Logistic Regression বা SVM) এই Data-তে খুব ভালো কাজ করবে।
    - Decision: যেই Feature-গুলো Data-কে সবচেয়ে ভালো আলাদা করতে পারছে, সেগুলোর উপর ফোকাস করা।

    সতর্কতা (Performance Issue):
    - Dataset-এ যদি অনেক বেশি Numerical Column (যেমন ১৫-২০টি) থাকে, তবে PairPlot লোড হতে অনেক সময় নেয় এবং গ্রাফগুলো খুব ছোট হয়ে যায়।
    - তখন পুরো Dataset একসাথে না দিয়ে, শুধু Important Column-গুলো Filter করে PairPlot তৈরি করা উচিত।
'''

pass
```

```python
'''
    PairPlot (tips dataset) থেকে ML Learner-এর Analysis ও Decision

    ১. Primary Diagonal (Distribution / Data-র বিস্তৃতি)
    ---------------------------------------------------
    Observation:
    - total_bill এবং tip-এর Distribution Right-skewed (ডানদিকে লম্বা লেজ)।
    - size Variable-টি Numerical হলেও এর মানগুলো Discrete (১, ২, ৩, ৪...)। ২ জনের গ্রুপ সবচেয়ে বেশি।
    - মনে রাখতে হবে: Diagonal-এর X-অক্ষে Feature-এর Value থাকে, আর Y-অক্ষে থাকে Count বা Frequency।

    ML Implication:
    - Highly skewed data কিছু Linear Model-এর (যেমন Linear Regression) Performance খারাপ করতে পারে।
    - size-কে Continuous Numerical হিসেবে না ধরে Ordinal Categorical Feature হিসেবে চিন্তা করা যেতে পারে।

    Decision:
    - Model train করার আগে skewed feature-গুলোর জন্য Log Transformation বা অন্য কোনো Scaling টেকনিক apply করার কথা ভাবতে হবে।


    ২. Feature vs Target Relationship (total_bill vs tip)
    -----------------------------------------------------
    Observation:
    - Off-diagonal scatter plot-এ total_bill এবং tip-এর মধ্যে স্পষ্ট Positive Correlation দেখা যাচ্ছে।
    - total_bill বাড়লে tip-ও বাড়ছে।

    ML Implication:
    - tip যদি আমাদের Target (Y) হয়, তবে total_bill একটি অত্যন্ত শক্তিশালী Predictive Feature (X)।

    Decision:
    - total_bill-কে অবশ্যই Model-এর জন্য Candidate Feature হিসেবে রাখতে হবে।


    ৩. Feature vs Feature Relationship (Multicollinearity Check)
    ------------------------------------------------------------
    Observation:
    - total_bill এবং size-এর মধ্যেও একটি পজিটিভ সম্পর্ক আছে (কাস্টমার বেশি হলে বিল বেশি হয়)।
    - তবে এটি একদম Perfect Linear সম্পর্ক নয়।

    ML Implication:
    - দুটি Independent Feature-এর মধ্যে সম্পর্ক থাকাকে Multicollinearity বলে।
    - Feature-গুলোর নিজেদের মধ্যে Correlation খুব বেশি হলে Model Unstable হতে পারে।

    Decision:
    - এখানে Correlation এতটাও স্ট্রং নয় যে কোনো একটি Feature বাদ দিতে হবে। তবে পরবর্তীতে Correlation Matrix (Heatmap) দিয়ে ভ্যালু চেক করে কনফার্ম হতে হবে।


    ৪. Next Step / করণীয়
    ---------------------
    - এই PairPlot-এ শুধু Numerical Column-গুলোর সম্পর্ক দেখা গেছে।
    - Categorical Data (যেমন: smoker, sex, time) Prediction-এ কোনো প্রভাব ফেলছে কিনা তা বুঝতে 'hue' parameter ব্যবহার করতে হবে।
    - Code: sns.pairplot(data=data, hue='time')

    Final Takeaway:
    ---------------
    PairPlot হলো পুরো Dataset-এর একটি "Bird's-eye view" বা এক্সরে রিপোর্ট।
    এটি দেখে একজন ML Engineer প্রাথমিক ধারণা পান যে:
    ১. কোন Feature-গুলো Target predict করতে ভালো কাজ করবে।
    ২. Data-তে কোনো সমস্যা (Skewness/Outlier) আছে কিনা।
    ৩. Model বানানোর আগে Data-তে কী ধরনের Preprocessing বা Transformation দরকার।
'''

pass
```
