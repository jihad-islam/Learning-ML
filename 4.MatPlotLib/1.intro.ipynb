```python
import numpy as np
```

```python
x = np.arange(5,20)
x
```

```python
y = x*5 +1
y
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(x,y)
```

```python
plt.plot(x,y,'red')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.title('jihads learning graph')
plt.show()
```

```python
# sub plot(multiple graph) creates

plt.subplot(1,2,1) # subplot(row,column, graph position)
plt.plot(x,y,'red')

plt.subplot(1,2,2)
plt.plot(y,x)

plt.show()
```

```python
# sub plot(multiple graph) creates

plt.subplot(1,2,1) # subplot(row,column, graph position)
plt.plot(x,y,'r--')

plt.subplot(1,2,2)
plt.plot(y,x)

plt.show()
```
