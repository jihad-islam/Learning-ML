```python
import matplotlib.pyplot as plt
import numpy as np
```

## histplot or histogram

```python
plt.style.use('_mpl-gallery') # eta matplotlib theke ekta built-in style input kora hoise. etar jonnei output er histrogram er background e graph style show kortese. ami chaile onno  graph e eta use korte parbo.  

# make data
np.random.seed(1) # seed diye ami fixed random array generate korlam.
x = 4 + np.random.normal(0, 1.5, 200)

# plot:
fig, ax = plt.subplots()

ax.hist(x, bins=8, linewidth=0.5, edgecolor="white") # histogram er jonne hist() plot use kora hoy

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),  # xlim vs xticks: xlim holo x axis er range, xticks holo x axis e kon kon value show korbe.
       ylim=(0, 56), yticks=np.linspace(0, 56, 9))

plt.show()


# code link: https://matplotlib.org/stable/plot_types/stats/hist_plot.html
```

```python
plt.style.use('_mpl-gallery')  

# make data
np.random.seed(1) 
x = 4 + np.random.normal(0, 1.5, 200) # imp: ei normal() method ta value gula emon vabe generate korbe jeno histogram er shape ta normal distribution er moto hoy.

# plot:
fig, ax = plt.subplots(figsize=(8, 6)) # graph er size change korte subplots() er moddhe figsize use korte hobe.

ax.hist(x, bins=8, linewidth=0.5, edgecolor="white") 

ax.set(xlim=(0, 8), xticks=np.arange(1, 8),  
       ylim=(0, 56), yticks=np.linspace(0, 56, 9))

plt.show()
```

## scatter plot

```python
plt.style.use('_mpl-gallery')

# make the data
np.random.seed(3)
x = 4 + np.random.normal(0, 2, 24)
y = 4 + np.random.normal(0, 2, len(x))
# size and color:
sizes = np.random.uniform(15, 80, len(x))
colors = np.random.uniform(15, 80, len(x))

# plot
fig, ax = plt.subplots(figsize=(10,5))

ax.scatter(x, y, s=sizes, c=colors, vmin=0, vmax=100)

ax.set(xlim=(0, 8), xticks=np.arange(1, 8), 
       ylim=(0, 8), yticks=np.arange(1, 8))

plt.show()
```
