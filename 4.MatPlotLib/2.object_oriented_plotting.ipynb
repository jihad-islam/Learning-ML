```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(5,20)
x
```

```python
y = x*5 +1
y
```

```python
# ekhan theke figure object pabe. jekhane size and axes pabo. axes holo graph er in details information.
fig2 = plt.figure() 
fig2
```

```python
axes = fig2.add_axes([.1, .1, .8,  .8]) #[left, bottom, width, height]
axes.plot(x,y)
```

```python
fig2
```

```python
axes2 = fig2.add_axes([.4, .2, .4, .5])

axes2.plot(x,y)
```

```python
fig2
```

```python
axes2.set_position([.2,.5,.3,.3])
fig2
```

```python
axes.set_xlabel('x axis')
axes.set_ylabel('y axis')
axes.set_title('jihads graph')
fig2
```
