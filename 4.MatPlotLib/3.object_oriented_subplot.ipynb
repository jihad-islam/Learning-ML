## legend() method

```python
fig = plt.figure()

axes = fig.add_axes([0,0,1,1]) # [left, bottom, width, height] in figure fraction
```

```python
axes.plot(x,y)
fig
```

```python
axes.plot(x,x**2)

fig
```

```python
# uporer graph e 2 ta line ase. kon line ta kisher jonne bojha jaitese na. eta identify kora hoy legend diye.

axes.plot(x,y, label='linear')
axes.plot(x,x**2, label='quadratic')

axes.legend(loc = 'upper left') # legend show korar jonno ei method use kora hoy. loc e legend er position set kora hoy. upper left, upper right, lower left, lower right, center etc.

fig
```
