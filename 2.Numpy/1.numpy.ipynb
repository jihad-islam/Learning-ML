### np.random.exponential()

```python
"""
    NumPy Random: rand() vs exponential()
    =====================================

    দুটো function-ই random value generate করে, কিন্তু তারা
    ভিন্ন distribution থেকে value generate করে।

    --------------------------------------------------------
    1. np.random.rand()
    --------------------------------------------------------

    rand() → Uniform distribution থেকে random value দেয়।

    Uniform distribution মানে:
    - 0 থেকে 1-এর মধ্যে value আসবে।
    - এই range-এর value-গুলো প্রায় সমানভাবে likely।

    Examples:

    np.random.rand()
    → ১টি random value
    → 0 ≤ value < 1

    np.random.rand(5)
    → ৫টি random value
    → সবগুলো 0 ≤ value < 1

    np.random.rand(10, 2)
    → 10 rows × 2 columns
    → মোট 20টি random value
    → সবগুলো 0 ≤ value < 1


    Important:
    rand() এ size= ব্যবহার করা যায় না।

    যেমন:
    np.random.rand(5)          # valid
    np.random.rand(size=5)     # invalid

    কারণ rand() function-এর syntax হলো:

    np.random.rand(d0, d1, d2, ...)

    অর্থাৎ এখানে সরাসরি dimensions/shape দিতে হয়।


    --------------------------------------------------------
    2. np.random.exponential()
    --------------------------------------------------------

    exponential() → Exponential distribution থেকে
    random value দেয়।

    এখানে value-এর fixed upper limit নেই।

    মূল rule:

    value ≥ 0

    অর্থাৎ:
    - negative value আসবে না।
    - 0 বা তার কাছাকাছি ছোট value বেশি likely।
    - বড় value আসতে পারে, কিন্তু বড় value-এর probability কম।
    - কোনো fixed maximum নেই।

    note: exponential distribution e 1-10 range dilam. ekhane 10 ashtei pare but tar probability kom. 1 er kache value asha beshi likely.
    graph:
                    frequency
                ↑
                |████████
                |██████
                |████
                |███
                |██
                |█
                +--------------------→ value
                    0   1   2   3   4...

    Examples:
    np.random.exponential()
    → ১টি random value

    np.random.exponential(size=5)
    → ৫টি random value

    np.random.exponential(size=(10, 2))
    → 10 rows × 2 columns
    → মোট 20টি random value


    এ ধরনের value আসতে পারে:

    [0.2, 0.7, 0.3, 1.8, 0.5, 4.2, 0.1, ...]

    এখানে বেশিরভাগ value ছোট হবে,
    মাঝে মাঝে বড় value আসতে পারে।


    --------------------------------------------------------
    3. rand() vs exponential()
    --------------------------------------------------------

    np.random.rand()

    → Uniform distribution
    → 0 ≤ value < 1
    → fixed range আছে
    → values প্রায় সমানভাবে distributed


    np.random.exponential()

    → Exponential distribution
    → value ≥ 0
    → fixed upper limit নেই
    → ছোট values বেশি likely
    → বড় values progressively কম likely


    --------------------------------------------------------
    4. সবচেয়ে সহজভাবে মনে রাখার নিয়ম
    --------------------------------------------------------

    rand()
    → "0 থেকে 1-এর মধ্যে random value"

    exponential()
    → "0 বা তার বেশি random value,
    যেখানে ছোট value বেশি এবং বড় value কম"


"""

pass
```

## shape manipulation

```python
# amra jokhon real project er dataset niye kaj korbo tokhon dataset er datagula kon dimension(row * column) e ase sheta directly dekhte parbo na. dimension dekhte shape attribute use korte hobe.

random_array = np.random.randint(10,21,[6,3,4]) # ekhane 10 to 21 er moddhe random 6x3x4 er 3d matrix pabo, 21 ta include hobe na.
random_array 
```

```python
random_array.shape  
```

```python
# reshape method diye amra kono array er shape change korte pari. but total number of element same thakbe. for example amader random_array er shape 6x3x4, mane total element 6*3*4=72. amra chaile 72 ta element ke 2d array te convert korte pari, for example 8x9 or 12x6 etc. but total element same thakbe.

# most important shape change korte hole total element number same thakte hobe.

new_shape_array = random_array.reshape(12,6) 

new_shape_array
```

```python
# find out the maximum and minimum value from the new_shape_array.

new_shape_array.max() # maximum value
```

```python
new_shape_array.min() # minimum value
```

```python
# maximum and minimum value er index find out korte hole amra argmax() and argmin() method use korte pari. 

# but argmax() and argmin() method ta 1D array er jonno kaj kore. tai amra new_shape_array ke 1D array te convert kore nibo.

new_shape_array_1d = new_shape_array.reshape(72) # ekhane 72 ta element ke 1D array te convert korlam.

new_shape_array_1d
```

```python
new_shape_array_1d.argmax() # maximum value er index
```

```python
new_shape_array_1d.argmin() # minimum value er index
```

### reshape(-1,1)

```python
"""
    reshape(-1, 1) মানে:
        -1 → কতটি row লাগবে, NumPy নিজে calculate করবে
        1 → 1টি column থাকবে

    Example:
        original_data.shape
        → (1000,)

        original_data.reshape(-1, 1)
        → (1000, 1)

    অর্থাৎ 1000টি element-কে 1টি column-এ সাজানো হলো।

    কেন reshape(1000, 1) এর বদলে reshape(-1, 1)?
        reshape(1000, 1)
        → 1000টি row hard-code করা হয়েছে।

        reshape(-1, 1)
        → row সংখ্যা NumPy নিজে বের করে।
        → data-এর size পরিবর্তন হলেও কাজ করবে।

    তাই reshape(-1, 1) বেশি flexible এবং practical।

    মনে রাখো:
        reshape(-1, 1)
        → "1টি column রাখো, row সংখ্যা তুমি নিজে বের করো।"
"""
```
