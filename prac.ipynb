```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Class": ["10", "10", "10", "9", "9", "9"],
    "Section": ["A", "A", "B", "A", "B", "B"],
    "Student": ["Rahim", "Karim", "Sakib", "Mina", "Rita", "Nabil"],
    "Marks": [80, 90, 70, 85, 75, 95]
})

print(df)
```

```python
df.groupby("Class")
```

```python
# upore ekta column diye group kora hosie. but dekhtei paccho table ta dekha jay na. amake dataframe er ekta object pathay.

# table ta dekhte hole amake kichu aggregation function use korle ekta visual table pabo.

# শুধু Class দিয়ে গ্রুপ, তারপর Marks এর যোগফল
df.groupby('Class')['Marks'].sum()
```

```python
df.groupby('Section')['Marks'].sum()
```

```python
# Class ও Section দিয়ে গ্রুপ
df.groupby(['Class', 'Section'])['Marks'].sum()
```

```python
# agg() দিয়ে count, sum, mean
df.groupby(['Class', 'Section'])['Marks'].agg(['count', 'sum', 'mean'])
```

```python
# section wise maximum marks student ber korbo.
max_id_by_section = df.groupby(['Section'])['Marks'].idxmax()
df.loc[max_id_by_section][['Section', 'Student', 'Marks']]
```

```python
# প্রতিটি গ্রুপের সর্বোচ্চ Marks-এর সারি বের করা
max_id_by_group = df.groupby(['Class', 'Section'])['Marks'].idxmax()
df.loc[max_id_by_group]
```
