### Choosing between loc and iloc (একটি গুরুত্বপূর্ণ পার্থক্য!)

`loc` এবং `iloc` এর মধ্যে কাজ করতে গেলে একটি বড় "gotcha" বা ফাঁদ আছে, যা মাথায় রাখা খুব জরুরি। এই দুটি মেথড ইনডেক্সিং করার সময় একটু ভিন্ন নিয়ম মেনে চলে।

* **`iloc`:** এটি Python-এর স্ট্যান্ডার্ড ইনডেক্সিং নিয়ম মেনে চলে। অর্থাৎ, রেঞ্জ দিলে প্রথমটি ইনক্লুড হয়, কিন্তু শেষেরটি এক্সক্লুড (বাদ) হয়ে যায়। তাই `0:10` দিলে এটি 0 থেকে 9 পর্যন্ত (মোট ১০টি) ভ্যালু সিলেক্ট করবে।
* **`loc`:** এটি **inclusively** (সবগুলো ধরে) ইনডেক্স করে। তাই `0:10` দিলে এটি 0 থেকে 10 পর্যন্ত (মোট ১১টি) ভ্যালু সিলেক্ট করবে!

**কেন এই পরিবর্তন?**
কারণ `loc` দিয়ে string বা টেক্সট ইনডেক্সও রেঞ্জ করা যায়। ধরুন, আপনার ইনডেক্সে 'Apples' থেকে 'Potatoes' পর্যন্ত অনেকগুলো নাম আছে। আপনি যদি `df.loc['Apples':'Potatoes']` লেখেন, তবে আপনি চাইবেন 'Potatoes' নামটিও লিস্টে থাকুক। যদি `iloc`-এর মতো কাজ করতো, তবে আপনাকে 'Potatoes'-এর পরের শব্দটি (যেমন 'Potatoet') খুঁজে বের করে লিখতে হতো, যা খুবই ঝামেলার।

এই নিয়মটি তখন খুব কনফিউজিং হয়ে যায়, যখন আপনার ইনডেক্সগুলো `0, 1, 2... 1000` নাম্বারে থাকে। `df.iloc[0:1000]` আপনাকে ঠিক 1000টি ডেটাই দেবে, কিন্তু `df.loc[0:1000]` আপনাকে 1001টি ডেটা দেবে! তাই `loc` দিয়ে ঠিক 1000টি ডেটা পেতে চাইলে আপনাকে এক ঘর কমিয়ে `df.loc[0:999]` লিখতে হবে।

## Conditional Selection

```python
data= np.random.randn(8, 5)

df = pd.DataFrame(
    data=data,
    index=['row1', 'row2', 'row3', 'row4', 'row5', 'row6', 'row7', 'row8'],
    columns=['col1', 'col2', 'col3', 'col4', 'col5']
)

df
```

```python
# jei value gula 0 er theke boro tader dekhte chai

df[df>0]
```

```python
# column 3 er jei value gula 0 er theke boro tader dekhte chai
df[df['col3']>0]
```

```python
# multiple conditional selection:

# pandas e multi conditional selection e ektu careful thakte hobe. python or cpp te amra contion apply kori logical and, logical or diye. but pandas e kintu bitwise and(&), bitwise or(|) use korte hobe.

# like ami column 3 and column 5 both column er jei value gula positive shudhu tader dekhabe.

df[(df['col3']>0) & (df['col5']>0)] # biwsie operator use korte must be bracket dite hobe condition e.
```

```python
# ami uporer code e  bitwise and(&) and use korsi. ami jodi logical and use kortam tahole error show korto.

df[(df['col3']>0) and (df['col5']>0)]
```

```python
# uporer error er reason holo logical and, logical or just boolean value niye kaj kore. but ekhane amra duita column er upor condition apply korsi. and ei column gula holo ekta series.

# so ekhane compare hocche: pandas series vs pandas series. tai amader bitwise operators use korte hobe.

type(df['col3']>0)
```
