```python
import numpy as np
import pandas as pd
```

# goupby() method

```python
np.random.seed(101)

data = {
    'Store Name': ['Pizza Hut', 'PizzaBurg', 'Dominos', 'Pizza Hut', 'Dominos', 'Pizza Hut', 'PizzaRella'],
    'Location': ['Dhanmondi', 'Taltola', 'Dhanmondi', 'Mirpur', 'Mirpur', 'Gulshan', 'taltola'],
    'Price' : np.random.randint(300, 700, size=7)
}
data
```

```python
df = pd.DataFrame(data)
df
```

```python
# goupby() er concept is from dbms

# goupby() method holo kono dataset er kono ekta column er basis e onno ekta column er upor aggregation operation perform kora.

# groupby() হলো একই ধরনের data-গুলোকে একসাথে group করা, তারপর সেই group-এর উপর কোনো calculation (aggregation) করা। 

# ki ki calculation kora jay groupby() method diye?: max, min, sum, mean, median, count, std, var etc. 

# aggregation operation gula usually numerical data er upor perform kora hoy. string er upor apply korle lexicographical order e kaj kore. jeta actully kono meaningless. 
```

```python
# ei code er output dekho ami actually price er sum jante chaisi but location ow lexicographically sum kore dise jeta meaningless.
# also amar dataset e kintu multiple dominos, pizza hut, etc ase. but groupby() method store name er upor apply koray same type er store ke eksathe rekhe dise.

df.groupby('Store Name').sum()
```

```python
# numeric_only= True  kore dile only numeric column gula niye kaj korbe.
df.groupby('Store Name').sum(numeric_only=True)
```

```python
# ami protita store er average price jante chai.
df.groupby('Store Name').mean(numeric_only=True) # ekhane numeric_only=True na dile error khabo.
```

```python
# ekta helpful method holo describe() method. eta diye amra easily kono column er summary statistics jante pari. and ei method by default numeric column er upor kaj kore.
df.groupby('Store Name').describe()
```
