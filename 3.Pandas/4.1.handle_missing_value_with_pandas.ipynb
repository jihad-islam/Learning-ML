````python
import pandas as pd
import numpy as np

# Sample Dataset
data = {
    "Name": ["Jihad", "Safia", "Rahim", "Karim"],
    "Department": ["CSE", np.nan, "EEE", "BBA"],
    "Score": [85, np.nan, 90, 78],
    "Age": [22, 21, np.nan, 23]
}

df = pd.DataFrame(data)

df
````

## 1. Missing Value খুঁজে বের করা

````python
# পুরো DataFrame-এ কোন cell-এ Missing Value আছে দেখো
df.isnull()
````

````python
# পুরো DataFrame-এ কোন cell Missing না
df.notnull()
````

### নির্দিষ্ট Column-এর Missing Value বের করা

````python
# Department column-এর যেসব row-তে NaN আছে
df[pd.isnull(df["Department"])]
````

````python
# Department column-এর যেসব row-তে NaN নেই
df[pd.notnull(df["Department"])]
````

## 2. প্রতিটি Column-এ কতগুলো Missing Value আছে

````python
# প্রতিটি column-এ কতটি NaN আছে
df.isnull().sum()
````

## 3. Missing Value Delete করা

````python
# NaN থাকা সব Row Delete

# যেসব row-তে অন্তত একটি NaN আছে, সেগুলো delete

df.dropna()
````

````python
# NaN থাকা সব Column Delete

# যেসব column-এ অন্তত একটি NaN আছে, সেগুলো delete
df.dropna(axis=1)
````

````python
'''
    - Column delete করা শেষ option।

    - শুধু NaN আছে বলে Column delete করা উচিত না।

    - প্রথমে দেখতে হবে:
        i) Column কতটা গুরুত্বপূর্ণ।
        ii) কত % Missing Value আছে।
        iii) অন্যভাবে Fill করা সম্ভব কিনা।
'''

pass
````

## 4. Missing Value Fill করা

````python
# সব NaN-এর জায়গায় "Missing" বসানো
df.fillna("Missing")
````

### নির্দিষ্ট Column Fill করা

````python
# Department column-এর NaN-এ Unknown বসানো
df.fillna({
    "Department": "Unknown"
})
````

````python
# Score column-এর NaN-এ 0 বসানো
df.fillna({
    "Score": 0
})
````

### একাধিক Column একসাথে Fill করা

````python
# Department এবং Score একসাথে Fill করা
df.fillna({
    "Department": "Unknown",
    "Score": 0
})
````

## 5. Mean দিয়ে Fill করা

````python
# Score column-এর Mean দিয়ে NaN Fill করা
df.fillna({
    "Score": df["Score"].mean()
})
````

````python
# Age column-এর Mean দিয়ে NaN Fill করা
df.fillna({
    "Age": df["Age"].mean()
})
````

### একাধিক Numerical Column Mean দিয়ে Fill

````python
# Score এবং Age-এর NaN Mean দিয়ে Fill
df.fillna({
    "Score": df["Score"].mean(),
    "Age": df["Age"].mean()
})
````

## 6. replace()

````python
# replace() Missing Value-এর জন্য না। এটা Existing Value পরিবর্তন করার জন্য।

# single value replace

# BBA কে Business দিয়ে Replace
df["Department"].replace(
    "BBA",
    "Business"
)
````

````python
# multiple value replace

# একাধিক Value একসাথে Replace
df.replace({
    "Department": {
        "BBA": "Business",
        "EEE": "Electrical"
    }
})
````

### Interview / Exam Important Methods

| Method           | কাজ                                     |
| ---------------- | --------------------------------------- |
| `isnull()`       | Missing Value খুঁজে বের করা             |
| `notnull()`      | Non-Missing Value খুঁজে বের করা         |
| `isnull().sum()` | প্রতিটি Column-এ কতটি Missing Value আছে |
| `dropna()`       | Missing Value থাকা Row/Column Delete    |
| `fillna()`       | Missing Value Fill করা                  |
| `replace()`      | Existing Value পরিবর্তন করা             |
