```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'col1': np.random.randint(1,10, size=10),
    'col2': np.random.randint(100,500, size=10),
    'col3': np.random.choice(['steve', 'jon', 'safia', 'jihad'], size=10)
})

df
```

## 4 mathods that always have to use of any dataset to get info

### 1. head() and tail() method

```python
# df.head()  it returns the first 5 rows of the dataframe
df.head()
```

```python
# df.head() ekhane amra chaile koita row dekhte chai taw define kore dite pari.
df.head(10)
```

```python
# df.tail()  it returns the last 5 rows of the dataframe
df.tail()
```

### 2. shape() method

```python
# df.shape  it returns the number of rows and columns in the dataframe
df.shape
```

### 3. info() method

```python
df.info()  # it returns the information about the dataframe
```

### 4. describe() method

```python
df.describe() # it returns the statistical summary of the dataframe
```

```python
df.describe(include='str')
```

## unique() and sort() method

### unique()

```python
# df.unique()  it returns the unique values in the dataframe
df['col3'].unique()
```

```python
# kotogula unique value ase sheta dekhete nunique() method use kora hoy
df['col3'].nunique()
```

```python
# kon unique value kotobar use kora hoise eta dekhte value_counts() method use kora hoy
df['col3'].value_counts()
```

### sort_values()

```python
# ami chai kono column er value ke sorted dekhte. sort_values() method use kora hoy ejonne.
df.sort_values(by='col2', ascending=False)  # it returns the dataframe sorted by col2 in descending order
```

## apply() and map() method

### apply() method - vvi

```python
# apply() method full dataset er upor customized function apply korte use kora hoy. and map() method specific column er upor customized function apply korte use kora hoy.

# map() diye ja ja kora jay tar shobgulai apply() diyei kora jay. but single column niye kaj korle convention holo map niye kaj kora but syntax mone na thakle obviously apply() use kora jabe. 
```

```python
# etokhon to in build method apply korlam. ami chaile amar nijer customized method o apply korte pari. and etar method holo apply() method.

# apply() method e ami jekono function ke argument hishebe pass korte pari. note: function call kora jabe na.

def my_func(x):
    return x*2

df['col3'].apply(my_func)
```

```python
# not only amar customized function, ami chaile python er built in function ow apply() method er argument hishebe pass korte pari.

# len()  holo python er built in function.  string er length ber kore dey.

df['col3'].apply(len)
```

### map() method

```python
# map() একটি column(series)-এর প্রতিটি value-এর ওপর একই function চালায় এবং একটি নতুন Series return করে।
```

```python
df = pd.DataFrame({
    "Name": ["Jihad", "Rahim", "Karim", "Safia"],
    "Marks": [80, 65, 90, 75]
})

df
```

```python
# ami marks column er protita value er sathe 5 add korbo

# ekhane lambda holo ekta function. python e function define korte jemon def keyword use kora hoy, tomoni map e lambda keyword or function ta use kora hoy.

# column e arithmatic operation apply korte lambda actually use kora hoy na. karon arithmatic operation aaro easy way tei apply kora jay. eta just example hishebe dekhano. map e amra ekta column er jeokno datatype er value er upor operation chalaite pari. tai map use kora hoy actually. 

df["Marks"].map(lambda x: x + 5)
```

```python
df["Marks"].map(lambda x: x * 2)
```

```python
df["Name"].map(
    lambda x: x + " Student"
)
```

```python
df["Name"].map(
    lambda x: x.upper()
)
```
