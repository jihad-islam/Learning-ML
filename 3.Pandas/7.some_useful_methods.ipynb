## apply() and map() method - vvi

```python
df2 = pd.DataFrame({
    "Name": ["Jihad", "Rahim", "Karim", "Safia"],
    "Marks": [80, 65, 90, 75],
    "Age": [23, 22, 24, 21]
})

df2
```

### apply() method

```python
# etokhon to in build method apply korlam. ami chaile amar nijer customized method o apply korte pari. and etar method holo apply() method.
# apply() method এর argument হিসেবে যেকোনো function pass করা যায়।
# Function CALL করা যাবে না, শুধু function এর নাম pass করতে হবে।

# apply() syntax: df['column_name'].apply(function_name)


def add_five(x):
    return x + 5

df2["Marks"].apply(add_five)
```

```python
# আরেকটি custom function

def make_upper(text):
    return text.upper()

df2["Name"].apply(make_upper)
```

```python
# Python এর built-in function ও apply() এর argument হিসেবে ব্যবহার করা যায়।

# len() প্রতিটি string এর length বের করে।

df2["Name"].apply(len)
```

```python
# apply method full dataset er uporew kaj korte pare jemon:

# প্রতিটি column এর maximum value বের করবে
df2.apply(max) # max() kintu ekta method. but apply() er moddhe call kora jabe na. just pass korlam.
```

### map() method

```python
# map() একটি column(series)-এর প্রতিটি value-এর ওপর একই function চালায় এবং একটি নতুন Series return করে।

# apply() and map() duitai same function duitari syntax same. just apply() is more powerful chaile full dataset er upor apply kora jabe, chaile single column er upor apply kora jabe. but map() is only for single column.

# map() syntax: df['column_name'].map(function_name)
```

```python
# নিজের (custom) function ব্যবহার

def add_five(x):
    return x + 5

df2["Marks"].map(add_five)
```

```python
# আরেকটি custom function

def make_upper(text):
    return text.upper()

df2["Name"].map(make_upper)
```

```python
# Python এর built-in function ব্যবহার

df2["Name"].map(len)
```

### lambda function in both map() and apply() method

```python
# lambda হলো Python-এর একটি anonymous (নামবিহীন) function।
# ছোট ও এক লাইনের logic লেখার জন্য lambda ব্যবহার করা হয়।
# lambda ব্যবহার করা বাধ্যতামূলক নয়। চাইলে একই কাজ def দিয়ে function লিখেও করা যায়।

# but lambda use korle code ta choto and precise hoy. alada kore r def function likha lage na.
```

```python
# lambda function in apply() method

df2["Marks"].apply(lambda x: x + 5)

df2["Marks"].apply(lambda x: x * 2)

df2["Name"].apply(lambda x: x + " Student")

df2["Name"].apply(lambda x: x.upper())

pass
```

```python
# lambda function in map() method

df2["Marks"].map(lambda x: x + 5)

df2["Marks"].map(lambda x: x * 2)

df2["Name"].map(lambda x: x + " Student")

df2["Name"].map(lambda x: x.upper())

pass
```

## sort method - sort_values()

```python
import pandas as pd

# Sample Dataset
df = pd.DataFrame({
    "Department": ["CSE", "EEE", "CSE", "BBA", "EEE", "BBA", "CSE", "EEE"],
    "Student": ["Rahim", "Mina", "Karim", "Hasan", "Nabil", "Sadia", "Sakib", "Rita"],
    "Marks": [85, 92, 92, 78, 85, 95, 78, 92],
    "Attendance": [90, 88, 95, 80, 91, 97, 85, 89]
})

df
```

### 1. Single Column Sort

```python
# Marks অনুযায়ী ছোট থেকে বড় (Ascending)
df.sort_values(by="Marks")
```

```python
# Marks অনুযায়ী বড় থেকে ছোট (Descending)
df.sort_values(by="Marks", ascending=False)
```

```python
# Student নাম Alphabetically সাজানো
df.sort_values(by="Student")
```

### 2. Multiple Column Sort

```python
# প্রথমে Marks অনুযায়ী বড় থেকে ছোট।
# যদি Marks সমান হয়, তাহলে Attendance অনুযায়ী ছোট থেকে বড়।
df.sort_values(
    by=["Marks", "Attendance"],
    ascending=[False, True]
)
```

```python
# প্রথমে Department অনুযায়ী Alphabetically।
# একই Department হলে Marks অনুযায়ী বড় থেকে ছোট।
df.sort_values(
    by=["Department", "Marks"],
    ascending=[True, False]
)
```

```python
# প্রথমে Marks অনুযায়ী ছোট থেকে বড়।
# Marks সমান হলে Student নাম অনুযায়ী Alphabetically।
df.sort_values(
    by=["Marks", "Student"],
    ascending=[True, True]
)
```
