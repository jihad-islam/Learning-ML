## merge method

````python
import numpy as np
import pandas as pd
````

````python
students = pd.DataFrame({
    "student_id": ["s1", "s2", "s3", "s4", "s5"],
    "name": ["Jihad", "Rahim", "Karim", "Nabil", "Asha"],
    "department_id": [101, 102, 103, 104, 105]
})

departments = pd.DataFrame({
    "department_id": [101, 102, 104, 106],
    "department_name": ["CSE", "EEE", "BBA", "English"],
    "building": ["A", "B", "C", "D"]
})
````

````python
students
````

````python
departments
````

````python
# inner join
students.merge(departments, on="department_id", how="inner") 
````

````python
# inner join ke evabew likha jay:

pd.merge(students, departments, on="department_id", how="inner")

# output same ashbe. but uporer ta beshi use kora hoy. uporer ta etari short form.
````

````python
# left join
# jei table theke dataframe call kora hoy oita always left table.

# students = left table, departments = right table
students.merge(departments, on="department_id", how="left")
````

````python
# right join

# students = left table, departments = right table
students.merge(departments, on="department_id", how="right")
````

````python
#outer join
students.merge(departments, on="department_id", how="outer")
````

````python
# join method amra khub beshi use korbo na. tai dekhar dorkar nai. join method diye ja korbo tar shobi amra merge method diye korte parbo.

# join er index:
students.join(
    departments.set_index("department_id"),
    on="department_id"
)
````

## Pandas: `join()` vs `merge()`

দুটোই একাধিক DataFrame একসাথে যুক্ত (combine) করার জন্য ব্যবহৃত হয়। পার্থক্য হলো **কোন ভিত্তিতে (key) DataFrame-গুলোকে মিলানো হচ্ছে।**

---

### `merge()`

- **Column-এর ভিত্তিতে** DataFrame যুক্ত করে।
- SQL-এর `JOIN`-এর মতো কাজ করে।
- Pandas-এ সবচেয়ে বেশি ব্যবহৃত হয়।

### Example

**students**

| ID | Name |
|---:|------|
| 1 | Jihad |
| 2 | Rahim |
| 3 | Karim |

**marks**

| ID | Marks |
|---:|------:|
| 1 | 90 |
| 2 | 85 |
| 3 | 95 |

```python
students.merge(marks, on="ID")
```

Result:

| ID | Name | Marks |
|---:|------|------:|
| 1 | Jihad | 90 |
| 2 | Rahim | 85 |
| 3 | Karim | 95 |

---

### `join()`

- **Index-এর ভিত্তিতে** DataFrame যুক্ত করে।
- সাধারণত দুই DataFrame-এর Index আগে থেকেই সম্পর্কিত থাকলে ব্যবহার করা হয়।

### Example

```python
students.join(marks)
```

যদি উভয় DataFrame-এর Index একই হয়।

Result:

| Index | Name | Marks |
|------:|------|------:|
| 0 | Jihad | 90 |
| 1 | Rahim | 85 |
| 2 | Karim | 95 |

---

### `merge()` vs `join()`

| Feature | `merge()` | `join()` |
|---------|-----------|----------|
| Match based on | Column | Index |
| Default key | Common column | Index |
| SQL JOIN-এর মতো | Yes | Similar |
| সবচেয়ে বেশি ব্যবহৃত | Yes | No |
| `set_index()` লাগে? | No | কখনও কখনও |
| Best use case | Common column থাকলে | Data আগে থেকেই indexed থাকলে |

---

### কখন কোনটি ব্যবহার করবে?

| পরিস্থিতি | Method |
|-----------|--------|
| দুই DataFrame-এ common column আছে | `merge()` |
| দুই DataFrame-এর Index একই | `join()` |
| SQL JOIN-এর মতো কাজ করতে চাও | `merge()` |
| Index align করে combine করতে চাও | `join()` |
