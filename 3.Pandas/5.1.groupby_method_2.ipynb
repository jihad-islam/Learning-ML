```python
import pandas as pd

# Sample Dataset
df = pd.DataFrame({
    "Department": ["CSE","CSE","CSE","EEE","EEE","EEE","BBA","BBA","BBA","CSE","EEE","BBA"],
    "Section":    ["A","A","B","A","A","B","A","B","B","B","B","A"],
    "Student":    ["Rahim","Karim","Sakib","Mina","Rita","Nabil","Hasan","Jony","Nila","Tanim","Farhan","Sadia"],
    "Marks":      [80,92,75,88,70,95,78,84,91,86,82,89],
    "Attendance": [90,95,82,88,80,97,85,87,94,91,86,93]
})

df
```

##  1.Create Group

```python
# শুধু group তৈরি করবে
df.groupby("Department")

# Output হবে DataFrameGroupBy object। কারণ এখনও কোনো operation করা হয়নি।
```

```python
# প্রতিটি group দেখতে
for name, group in df.groupby("Department"):
    print(name)
    print(group)
```

## 2. single column grouping operations

```python
# Department-wise মোট Marks
df.groupby("Department")["Marks"].sum()

# Department-wise Average Marks
df.groupby("Department")["Marks"].mean()

# Department-wise Highest Marks
df.groupby("Department")["Marks"].max()

# Department-wise Lowest Marks
df.groupby("Department")["Marks"].min()

# Department-wise Student Count
df.groupby("Department")["Marks"].count()

pass
```

```python
df.groupby("Department")["Marks"].sum()
```

## 3. Multiple Column Grouping

```python
# Department + Section অনুযায়ী Marks এর Sum
df.groupby(["Department", "Section"])["Marks"].sum()
```

## 4. Multiple Aggregation

```python
# একই সাথে একাধিক statistics
df.groupby("Department")["Marks"].agg([
    "count",
    "sum",
    "mean",
    "min",
    "max"
])
```

## 5. Multiple Numeric Columns

```python
# Marks এবং Attendance এর Average
df.groupby("Department")[["Marks", "Attendance"]].mean()
```

## 6. Highest Marks Student

```python
# প্রতিটি Department-এর Highest Marks Student-এর index
idx = df.groupby("Department")["Marks"].idxmax()

# সেই row গুলো বের করা
df.loc[idx][["Department", "Student", "Marks"]]
```

## 7. Highest Student (Multiple Group)

```python
# Department + Section অনুযায়ী Highest Marks Student
idx = df.groupby(["Department", "Section"])["Marks"].idxmax()

df.loc[idx]
```

## 8. Useful Methods

```python
# প্রতিটি group-এ মোট row
df.groupby("Department").size()

# নির্দিষ্ট group দেখা
df.groupby("Department").get_group("CSE")

# Group অনুযায়ী index
df.groupby("Department").groups

# MultiIndex কে normal DataFrame বানানো
df.groupby(["Department", "Section"])["Marks"].sum().reset_index()

# Average Marks অনুযায়ী Sort
df.groupby("Department")["Marks"].mean().sort_values(ascending=False)

pass
```

## 9. groupby + apply method

```python
# প্রতিটি Department থেকে Top 2 Marks পাওয়া Student বের করা
df.groupby("Department").apply(
    lambda group: group.nlargest(2, "Marks")
)
```

```python
# প্রতিটি Department-এ যাদের Marks Department-এর Average-এর চেয়ে বেশি
df.groupby("Department").apply(
    lambda group: group[group["Marks"] > group["Marks"].mean()]
)
```
