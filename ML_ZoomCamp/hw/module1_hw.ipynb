# Q7. Sum of Weights

1. Select all the cars from **Asia**.
2. Select only the columns `vehicle_weight` and `model_year`.
3. Select the **first 7 values**.
4. Get the underlying **NumPy array**. Let's call it `X`.
5. Compute matrix-matrix multiplication between the **transpose of `X`** and `X`. To get the transpose, use `X.T`. Let's call the result `XTX`.
6. **Invert `XTX`**.
7. Create an array `y` with values:
   ```python
   [1100, 1300, 800, 900, 1000, 1100, 1200]


````python
data = pd.read_csv('car_fuel_efficiency_2026.csv')

data.head()
````

````python
asian_data = data[data['origin'] == 'Asia']

asian_data.head()
````

````python
sliced_data =asian_data[['vehicle_weight', 'model_year']].iloc[0:7]
sliced_data
````

````python
x = sliced_data.to_numpy()
````

````python
xtx = x.T @ x

xtx
````

````python
xtx_inv =np.linalg.inv(xtx)

xtx_inv
````

````python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

y
````

````python
w = xtx_inv @ x.T @ y

w
````

````python
w.sum()
````
