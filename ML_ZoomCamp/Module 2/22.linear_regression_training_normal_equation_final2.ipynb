# Linear Regression Training — Normal Equation

## Goal

আগের lesson-এ আমরা জানতাম:

$$
\hat{y}=Xw
$$

অর্থাৎ `X` এবং `w` জানা থাকলে prediction বের করতে পারি।

এখন আসল প্রশ্ন:

> **এই weights `w` কোথা থেকে আসে?**

এই lesson-এ আমরা **Normal Equation** ব্যবহার করে Linear Regression-এর weights বের করব।

---

## Main Idea

আমাদের কাছে আছে:

- `X` = feature matrix
- `y` = actual target values
- `w` = unknown weights

আমরা চাই:

$$
\boxed{Xw\approx y}
$$

অর্থাৎ model-এর predictions যেন actual target-এর যতটা সম্ভব কাছে যায়।

Normal Equation:

$$
\boxed{w=(X^TX)^{-1}X^Ty}
$$

এখানে:

$$
X^TX
$$

কে **Gram Matrix** বলা হয়।

# 1. কেন সরাসরি `X` inverse করা যাবে না?

প্রথমে খুব simpleভাবে ভাবি:

$$
Xw=y
$$

যদি `X` invertible হতো, তাহলে দুই পাশে $X^{-1}$ দিয়ে multiply করতে পারতাম:

$$
X^{-1}Xw=X^{-1}y
$$

কারণ:

$$
X^{-1}X=I
$$

তাই:

$$
Iw=X^{-1}y
$$

এবং:

$$
\boxed{w=X^{-1}y}
$$

কিন্তু Linear Regression dataset-এ `X` সাধারণত **square matrix নয়**।

---

## Example

ধরি আমাদের dataset-এ:

- `m = 9`টি car
- `n = 3`টি feature

তাহলে:

$$
X\text{ shape}=9\times3
$$

অর্থাৎ:

```text
9 rows    → 9 cars
3 columns → 3 features
```

এটি square নয়, কারণ:

$$
9\neq3
$$

আর সাধারণ matrix inverse-এর জন্য matrix-কে square হতে হয়।

তাই:

$$
X^{-1}
$$

এখানে ব্যবহার করা যাবে না।

**এই কারণেই আমরা অন্য একটি পথ নিই।**

# 2. `Xw ≈ y` থেকে Gram Matrix

আমাদের মূল relation:

$$
Xw\approx y
$$

এখন দুই পাশে $X^T$ দিয়ে multiply করি:

$$
X^TXw\approx X^Ty
$$

Lesson-এ এটাকে shorthand করে লেখা হয়:

$$
\boxed{X^TXw=X^Ty}
$$

এখন লক্ষ্য করো:

$$
X^TX
$$

এর shape কী হবে?

যদি:

$$
X=(m\times n)
$$

তাহলে:

$$
X^T=(n\times m)
$$

তাই:

$$
X^TX=(n\times m)(m\times n)
$$

ফলে:

$$
\boxed{X^TX=n\times n}
$$

অর্থাৎ `X` rectangular হলেও `X.T @ X` একটি **square matrix**।

এই matrix-টাই:

## Gram Matrix

$$
\boxed{G=X^TX}
$$

আমাদের ক্ষেত্রে `X` যদি `9 × 3` হয়, তাহলে:

```text
X     = 9 × 3
X.T   = 3 × 9
X.T X = 3 × 3
```

এখন `X.T @ X`-এর inverse নেওয়ার possibility তৈরি হলো।

# 3. Normal Equation Derivation

আমরা পেয়েছি:

$$
X^TXw=X^Ty
$$

এখন দুই পাশে $(X^TX)^{-1}$ দিয়ে multiply করি:

$$
(X^TX)^{-1}X^TXw
=
(X^TX)^{-1}X^Ty
$$

কারণ:

$$
(X^TX)^{-1}(X^TX)=I
$$

তাই left side:

$$
Iw=w
$$

এবং আমরা পাই:

$$
\boxed{
w=(X^TX)^{-1}X^Ty
}
$$

এটাই **Normal Equation**।

### খুব গুরুত্বপূর্ণ

এই `w` original:

$$
Xw=y
$$

system-এর exact solution নয়, কারণ সাধারণত exact solution exist করে না।

বরং এটি এমন একটি `w` যা predictions:

$$
Xw
$$

কে actual `y`-এর **সবচেয়ে কাছাকাছি possible solution** দেয়।

অর্থাৎ:

$$
\boxed{Xw\approx y}
$$

# 5. Step 1 — Gram Matrix

Normal Equation:

$$
w=(X^TX)^{-1}X^Ty
$$

প্রথম কাজ:

$$
X^TX
$$

Python:

```python
XTX = X.T.dot(X)
```

Dimension:

$$
X=(9\times3)
$$

$$
X^T=(3\times9)
$$

তাই:

$$
X^TX=(3\times9)(9\times3)
=(3\times3)
$$

অর্থাৎ rectangular `X` থেকে square Gram Matrix তৈরি হলো।

````python
XTX = X.T.dot(X)

print("XTX shape:", XTX.shape)
print("\nGram matrix X.T @ X =")
print(XTX)
````

# 6. Step 2 — Gram Matrix-এর Inverse

এখন:

$$
(X^TX)^{-1}
$$

Python:

```python
XTX_inv = np.linalg.inv(XTX)
```

### কেন inverse নিচ্ছি?

কারণ Normal Equation-এর derivation-এ:

$$
X^TXw=X^Ty
$$

থেকে `w` isolate করতে inverse ব্যবহার করছি:

$$
w=(X^TX)^{-1}X^Ty
$$

### Sanity Check

যদি inverse সঠিক হয়:

$$
(X^TX)(X^TX)^{-1}=I
$$

Identity matrix-এর diagonal-এ `1` এবং অন্য জায়গায় `0` থাকে।

Floating-point calculation-এর কারণে exact `0` এর বদলে খুব ছোট number দেখা যেতে পারে, যেমন `-0.0`। তাই `.round(1)` ব্যবহার করে সহজে যাচাই করা যায়।

````python
XTX_inv = np.linalg.inv(XTX)

identity_check = XTX.dot(XTX_inv).round(1)

print("XTX @ XTX_inv =")
print(identity_check)
````

# 9. Step 5 — Calculate the Weights

এখন আমাদের দুইটি জিনিস already আছে:

### কিন্তু একটা বিষয় এখনও বাকি

এখন পর্যন্ত `X`-এ **bias column নেই**।

অর্থাৎ আমরা এমন model বানিয়েছি:

$$
\hat y=w_1x_1+w_2x_2+w_3x_3
$$

কিন্তু Linear Regression-এ সাধারণত bias দরকার:

$$
\hat y=w_0+w_1x_1+w_2x_2+w_3x_3
$$

তাই এখন bias add করব।

# 10. Bias Term কেন দরকার?

আগের lesson থেকে মনে করো:

$$
\hat y=w_0+x_1w_1+x_2w_2+x_3w_3
$$

Bias $w_0$ হলো baseline বা intercept।

Bias-কে matrix multiplication-এর মধ্যে রাখার জন্য প্রতিটি row-এর শুরুতে `1` বসাই:

```text
[1, x1, x2, x3]
```

তাহলে:

$$
1\times w_0+x_1w_1+x_2w_2+x_3w_3
$$

যেহেতু:

$$
1\times w_0=w_0
$$

তাই original Linear Regression formula-ই পাওয়া যায়।

# 13. এখন আবার Normal Equation চালাই

Bias add করার পরে নতুন `X`:

$$
X:9\times4
$$

তাই:

$$
X^T:4\times9
$$

এখন:

$$
X^TX=(4\times9)(9\times4)
=4\times4
$$

এবং:

$$
X^Ty=(4\times9)(9\times1)
=4\times1
$$

তারপর:

$$
w=(X^TX)^{-1}X^Ty
$$

এবার `w`-এর মধ্যে মোট 4টি value থাকবে:

$$
w=
\begin{bmatrix}

w_2\\

$$

````python
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)

w_full = XTX_inv.dot(X.T).dot(y)

print("XTX shape:", XTX.shape)
print("w_full shape:", w_full.shape)

print("\nFull weight vector:")
print(w_full)
````

এটা dataset এবং feature-এর context-এর ওপর নির্ভর করে interpret করতে হবে।

````python
w0 = w_full[0]
w = w_full[1:]

print("Bias (w0):", w0)
print("Feature weights (w):", w)
````

# 16. পুরো Normal Equation Code

এখন সব step একসাথে:

```python
ones = np.ones(X.shape[0])
X = np.column_stack([ones, X])

XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)

w_full = XTX_inv.dot(X.T).dot(y)

w0 = w_full[0]
w = w_full[1:]
```

এখানে লক্ষ্য রাখো:

> **Bias column add করার দায়িত্বও function-এর ভিতরে রাখা যায়।**

তাহলে user শুধু normal feature matrix এবং target `y` পাঠাবে।

````python
def train_linear_regression(X, y):
    # Add bias column
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    # Normal Equation
    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)

    # Split bias and feature weights
    return w_full[0], w_full[1:]
````

# 17. Function ব্যবহার

এখন function-এ **bias column ছাড়া original feature matrix** এবং `y` দেব।

Function নিজেই:

1. bias column add করবে
2. Gram Matrix তৈরি করবে
3. Gram Matrix-এর inverse বের করবে
4. $X^Ty$ বের করবে
5. Normal Equation দিয়ে weights বের করবে
6. bias এবং feature weights আলাদা করে return করবে

````python
# Original feature matrix without the bias column
X_features = np.array([
    [148, 24, 1385],
    [132, 25, 2031],
    [453, 11, 86],
    [158, 24, 185],
    [172, 25, 201],
    [413, 11, 86],
    [38,  54, 185],
    [142, 25, 431],
    [453, 31, 86],
], dtype=float)

y = np.array([
    100, 200, 150, 250, 100,
    200, 150, 250, 120
], dtype=float)

w0, w = train_linear_regression(X_features, y)

print("Bias w0:", w0)
print("Feature weights w:", w)
````

# 18. `X` rectangular কিন্তু `X.T @ X` square — এটা মনে রাখো

এটাই এই lesson-এর সবচেয়ে important matrix concept।

ধরি:

$$
X=m\times n
$$

যেখানে:
- `m` = examples/rows
- `n` = features/columns

তাহলে:

$$
X^T=n\times m
$$

এবং:

$$
X^TX=(n\times m)(m\times n)
$$

সুতরাং:

$$
\boxed{X^TX=n\times n}
$$

Example:

```text
X       = 9 × 3
X.T     = 3 × 9
X.T @ X = 3 × 3
```

Bias add করলে:

```text
X       = 9 × 4
X.T     = 4 × 9
X.T @ X = 4 × 4
```

এই square matrix-টিই হলো **Gram Matrix**।
