# Linear Regression — Vector Form

## Core Idea

Linear Regression-এর prediction ধাপে ধাপে এভাবে compact করা যায়:

**একটি car → Dot Product → সব car → Matrix × Vector**

### 1. One car

$$
g(x_i)=w_0+\sum_{j=1}^{n}x_{ij}w_j
$$

এখানে:
- $x_i$ = একটি car-এর feature vector
- $w$ = weights
- $w_0$ = bias
- $g(x_i)$ = prediction

### 2. Dot Product

$$
\sum x_{ij}w_j = x_i \cdot w
$$

তাই,

$$
g(x_i)=w_0+x_i\cdot w
$$

### 3. Fake / Fictional Feature

Bias-কে dot product-এর ভিতরে নেওয়ার জন্য নতুন feature নিই:

$$
x_{i0}=1
$$

কারণ,

$$
1\times w_0=w_0
$$

তাই feature vector এবং weight vector হয়:

$$
x_i=[1,x_{i1},x_{i2},...,x_{in}]
$$

$$
w=[w_0,w_1,w_2,...,w_n]
$$

এখন:

$$
g(x_i)=x_i\cdot w
$$

### 4. All cars

সব car-এর feature vector একসাথে matrix $X$ করলে:

$$
\boxed{\hat y=Xw}
$$

অর্থাৎ **Linear Regression prediction = Matrix × Weight Vector**.

## Example — One Car

ধরি:

- Features = `[148, 24, 1385]`
- Bias = `7.17`
- Weights = `[0.01, 0.04, 0.002]`

তাহলে:

$$
g(x)=7.17+(148)(0.01)+(24)(0.04)+(1385)(0.002)
$$

$$
=7.17+1.48+0.96+2.77
$$

$$
=\boxed{12.38}
$$

Fake feature ব্যবহার করলে:

$$
[1,148,24,1385]\cdot[7.17,0.01,0.04,0.002]
=12.38
$$

````python
# Dot product using a simple loop

def dot(xi, w):
    n = len(xi)
    res = 0.0

    for j in range(n):
        res = res + xi[j] * w[j]

    return res


# Example
xi = [148, 24, 1385]
w = [0.01, 0.04, 0.002]

print(dot(xi, w))
````

````python
# Linear Regression for one car

w0 = 7.17
w = [0.01, 0.04, 0.002]

def linear_regression_one_car(xi):
    xi = [1] + xi
    w_new = [w0] + w

    return dot(xi, w_new)


xi = [148, 24, 1385]

print(linear_regression_one_car(xi))
````

## 4. All Cars — Matrix × Vector

এখন পর্যন্ত আমরা **একটা car**-এর prediction বের করেছি।  
বাস্তব dataset-এ অনেকগুলো car থাকে, তাই একই calculation সব car-এর জন্য একসাথে করতে চাই।

### Step 1 — তিনটি car

**Car 1**

$$
x_1=[1,148,24,1385]
$$

**Car 2**

$$
x_2=[1,132,25,2031]
$$

**Car 3**

$$
x_3=[1,453,11,86]
$$

এখানে প্রথম `1` হলো **fictional feature**, যেটা bias $w_0$-কে dot product-এর মধ্যে আনে।

### Step 2 — সব car-কে একটি Matrix-এ রাখি

$$
X=
\begin{bmatrix}
1&148&24&1385\\
1&132&25&2031\\
1&453&11&86
\end{bmatrix}
$$

এখানে:

- **3 rows** = 3টি car
- **4 columns** = bias + 3টি features

তাই:

$$
X\text{ এর shape}=3\times4
$$

### Step 3 — Weight Vector

$$
w=
\begin{bmatrix}
7.17\\
0.01\\
0.04\\
0.002
\end{bmatrix}
$$

তাই:

$$
w\text{ এর shape}=4\times1
$$

### Step 4 — Matrix × Vector

এখন:

$$
Xw=
\begin{bmatrix}
1&148&24&1385\\
1&132&25&2031\\
1&453&11&86
\end{bmatrix}
\begin{bmatrix}
7.17\\
0.01\\
0.04\\
0.002
\end{bmatrix}
$$

এখানে **প্রতিটি row-এর সাথে `w`-এর dot product হচ্ছে**।

অর্থাৎ:

$$
\text{Row 1}\cdot w \rightarrow \text{Prediction 1}
$$

$$
\text{Row 2}\cdot w \rightarrow \text{Prediction 2}
$$

$$
\text{Row 3}\cdot w \rightarrow \text{Prediction 3}
$$

### Step 5 — প্রতিটি prediction

#### Car 1

$$
[1,148,24,1385]\cdot[7.17,0.01,0.04,0.002]
$$

$$
=(1)(7.17)+(148)(0.01)+(24)(0.04)+(1385)(0.002)
$$

$$
=7.17+1.48+0.96+2.77
=\boxed{12.38}
$$

#### Car 2

$$
[1,132,25,2031]\cdot[7.17,0.01,0.04,0.002]
$$

$$
=(1)(7.17)+(132)(0.01)+(25)(0.04)+(2031)(0.002)
$$

$$
=7.17+1.32+1+4.062
=\boxed{13.552}
$$

#### Car 3

$$
[1,453,11,86]\cdot[7.17,0.01,0.04,0.002]
$$

$$
=(1)(7.17)+(453)(0.01)+(11)(0.04)+(86)(0.002)
$$

$$
=7.17+4.53+0.44+0.172
=\boxed{12.312}
$$

### Final result

তাই:

$$
Xw=
\begin{bmatrix}
12.38\\
13.552\\
12.312
\end{bmatrix}
$$

এই vector-টাই আমাদের **predictions vector**:

$$
\boxed{\hat y=Xw}
$$

### সবচেয়ে গুরুত্বপূর্ণ idea

`X.dot(w)` কোনো নতুন ML formula নয়।

এটা শুধু:

> **X-এর প্রতিটি row-এর সাথে w-এর dot product একসাথে করে।**

তাই:

```text
X-এর row 1 → dot(w) → prediction 1
X-এর row 2 → dot(w) → prediction 2
X-এর row 3 → dot(w) → prediction 3
```

Dataset-এ যদি 11,914টি car থাকে, তাহলে একই operation একবার চালিয়েই 11,914টি prediction পাওয়া যাবে।

````python
import numpy as np

# Three cars
x1  = [1, 148, 24, 1385]
x2  = [1, 132, 25, 2031]
x10 = [1, 453, 11, 86]

# Feature matrix
X = np.array([x1, x2, x10])

# Bias + weights
w0 = 7.17
w = [0.01, 0.04, 0.002]
w_new = [w0] + w

# Linear Regression for all cars
def linear_regression(X):
    return X.dot(w_new)

predictions = linear_regression(X)

print(predictions)
````
